# 0. Red Neuronal con Tensorflow - Grados Celcius a Farenheit

```python
# 0.1 Importamos los datos del archivo
import pandas as pd
datos = pd.read_csv("datos.csv")
```

```python
# 0.2 Vemos los datos del archivo
datos.head()
```

# 1. Preparamos los datos de entrada y salida

```python
# 1. Preparamos los datos de Entrada (X) y Salida (Y)
X = datos["celcius"].values
Y = datos["farenheit"].values
```

# 2. Preparamos nuestro modelo en tensorflow

```python
# 2.1 Importamos tensorflow
import tensorflow as tf

modelo = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(1,)), # con esto especificamos que el perceptron tiene 1 sola entrada los grados celcius
        tf.keras.layers.Dense(units=1) # Especificamos el numero de neuronas que queremos Salida : Grados Farenheit
])
```

```python
# 2.2 Compilamos el modelo
modelo.compile(
    optimizer=tf.keras.optimizers.Adam(.1),  # indica el optimizador con su taza de aprendizaje = .1
                                            # Mas pequenio sea la taza de optimizador la red tarda muchas epocas en aprender
                                            # Mas Grande , tarda mas.
    loss="mean_squared_error"          # La perdida que va a medir.
)
```

# 3. Entrenamos la Red Neural

```python
# 3. Entrenamos la red
entrenamiento = modelo.fit(X, # Las entradas
                           Y, # Las Salidas
                           epochs=1000 # Numero de epocas cuantas vueltas completas dara al set de datos
                                       # En cada epoca, va a ver que tan mal le fue con el loss = "mean_squared_error"
                                       # y lo optimizara .1 su taza de aprendizaje, con cada set de datos 1000 Veces
                           )
```

```python
# 3.1 Hacemos una grafica para ver las perdida en cada epoca
import matplotlib.pyplot as plt
plt.xlabel(" # Expoca ")
plt.ylabel(" Magnitud de la perdida")
plt.plot(entrenamiento.history["loss"])
```

# 4. Hacemos predicciones

```python
# 4.1 Realizamos una prediccion
import numpy as np
c = np.array([1230])
prediccion = modelo.predict([c])
print(f"Prediccion: {c} celcius son {prediccion[0]} farenheit")
```

# 0 Analizamos datos con la Maquina de Vectores y De Soportes

```python
# Estos sirven para analizar pocos datos con muchas caracteristicas

# El modelo SVM (Support Vector Machine, Máquina de Vectores de Soporte) es un algoritmo de aprendizaje supervisado utilizado principalmente para problemas de clasificación y regresión. Su objetivo es encontrar el hiperplano que mejor separa las diferentes clases en el espacio de características. Aquí hay algunas aplicaciones comunes de SVM:

# - Clasificación binaria: SVM se utiliza comúnmente para clasificar datos en dos clases. Por ejemplo, en problemas de detección de spam, SVM puede usarse para clasificar correos electrónicos como spam o no spam.
# - Clasificación multiclase: SVM también se puede extender para clasificar datos en más de dos clases. Por ejemplo, en reconocimiento de dígitos manuscritos, SVM puede usarse para clasificar los dígitos del 0 al 9.
# - Clasificación no lineal: SVM puede manejar eficazmente problemas de clasificación no lineales mediante el uso de funciones de kernel, que transforman los datos en un espacio dimensional superior donde es más fácil encontrar un hiperplano de separación lineal.
# - Regresión: Aunque SVM es más conocido por la clasificación, también puede utilizarse para problemas de regresión. En este caso, SVM encuentra un hiperplano que mejor se ajusta a los datos de entrenamiento de una manera que minimiza el error de regresión.
# - Detección de anomalías: SVM se puede utilizar para detectar anomalías en conjuntos de datos, como fraudes en transacciones financieras o fallos en sistemas.
# En resumen, SVM es un modelo versátil que se puede utilizar en una variedad de problemas de aprendizaje supervisado, especialmente cuando se trata de problemas de clasificación lineal o no lineal.
```

# 1 Importamos los datos 

```python
# 1.- IMportamos los datos
from sklearn import datasets
datos = datasets.load_breast_cancer(as_frame=True # Colocamos as_frame=True para cargarlo como ser de datos
                                    )
```

# 3. Separamos los datos de entradas y salidas para preparar el modelo

```python
# 3.1 Separamos los datos
from sklearn.model_selection import train_test_split

x = datos.data
y = datos.target
x_ent, x_pruebas, y_ent, y_pruebas = train_test_split(x,y,test_size= .3)
```

# 4. Creamos el modelo SVM y probamos los Kernels y Gamas

```python
# 4.1 Importamos el modelo Maquinas de vectores
from sklearn import svm
from sklearn import metrics


# 4.2 Usamos el modelo SVC Supported Vector Classifier
modelo = svm.SVC() 

# 4.3 Entrenamos el modelo
modelo.fit(x_ent,y_ent)

# 4.4 Hacemos unas predicciones
pedicciones = modelo.predict(x_pruebas)

print("Exactitud:", metrics.accuracy_score(y_pruebas, pedicciones))
print("Precicion:", metrics.precision_score(y_pruebas, pedicciones))
```

```python
# 4.5 Modificamos el codigo anterior para probar con los 3 Kernels
# y los valores de Gama para saber cual da mejor exactitud y precision
# 4.1 Importamos el modelo Maquinas de vectores
from sklearn import svm
from sklearn import metrics

# 4.1.1 Ponemos los kernels que transforman los datos para probarlos
kernels = ["linear","rbf","sigmoid"]
gamas = [1, .01, .001, .0001]

for kernel in kernels:
    for gama in gamas:

        # 4.2 Usamos el modelo SVC Supported Vector Classifier
        modelo = svm.SVC(kernel=kernel, gamma = gama) 

        # 4.3 Entrenamos el modelo
        modelo.fit(x_ent,y_ent)

        # 4.4 Hacemos unas predicciones
        pedicciones = modelo.predict(x_pruebas)

        print(f"Kernel {kernel} & Gama {gama} -> Exactitud:", metrics.accuracy_score(y_pruebas, pedicciones))
        print(f"Kernel {kernel} & Gama {gama} -> Precicion:", metrics.precision_score(y_pruebas, pedicciones))

```

```python
#NOTA : COn esto podemos ver que al kernel que mejor deio Precision y Exactitud
# Fue el "Lineal" y al que peor le fue con el conjunto de datos gue el Sigmoid
# por lo tanto lo usamos para nuestro modelo 

# 4.2 Usamos el modelo SVC Supported Vector Classifier
modelo = svm.SVC(kernel="linear") 

# 4.3 Entrenamos el modelo
modelo.fit(x_ent,y_ent)

# 4.4 Hacemos unas predicciones
pedicciones = modelo.predict(x_pruebas)

print(f"Kernel {kernel} & Gama {gama} -> Exactitud:", metrics.accuracy_score(y_pruebas, pedicciones))
print(f"Kernel {kernel} & Gama {gama} -> Precicion:", metrics.precision_score(y_pruebas, pedicciones))
```

# 5.- Hacemos el reporte de clasificacion

```python
# 5.1 Importamos las librerias
from sklearn.metrics import classification_report

print(classification_report(y_pruebas, pedicciones))
```

```python
# 5.2 Sacamos la matriz de confusion
from sklearn.metrics import confusion_matrix
import pandas as pd

pd.DataFrame( confusion_matrix(y_pruebas, pedicciones))
```
